7. Understand which campaigns had the most successful outcomes. Compare the total quantities sold before and after the promotions for the Sankranti and Diwali campaigns. Which campaign saw a greater increase in sales?
- Sankranti campaign saw greater increase in sales.

```python
df_merged['revenue_before_promo'] = df_merged['base_price(before_promo)'] * df_merged['quantity_sold(before_promo)']
df_merged['revenue_after_promo'] = df_merged['base_price(after_promo)'] * df_merged['quantity_sold(after_promo)']
```

```python
df_merged['IR']=(df_merged['revenue_after_promo']-df_merged['revenue_before_promo'])/df_merged['revenue_before_promo']*100
```

```python
df_merged.head()
```

```python
sankranti_df = df_merged[df_merged['campaign_name'] == 'Sankranti'].copy()

product_revenue = sankranti_df.groupby('product_code').agg(
    revenue_before=('revenue_before_promo', 'sum'),
    revenue_after=('revenue_after_promo', 'sum')
).reset_index()

product_revenue['IR%'] = ((product_revenue['revenue_after'] -product_revenue['revenue_before'])/ product_revenue['revenue_before']) * 100

result = product_revenue.sort_values('IR%', ascending=False).head(1)

result
```

9. Which store in Visakhapatnam recorded the lowest Incremental Sold Units Percentage (ISU%) during the Diwali campaign? What is the ISU% for that store?
    - STVSK-3,49.21


```python
sankranti_df = df_merged[df_merged['campaign_name'] == 'Sankranti'].copy()

promo_summary = sankranti_df.groupby('promo_type').agg(
    revenue_before=('revenue_before_promo', 'sum'),
    revenue_after=('revenue_after_promo', 'sum'),
    units_before=('quantity_sold(before_promo)', 'sum'),
    units_after=('quantity_sold(after_promo)', 'sum')
)

promo_summary['IR%'] = ((promo_summary['revenue_after'] - promo_summary['revenue_before'])/ promo_summary['revenue_before']) * 100

promo_summary['ISU%'] = ((promo_summary['units_after'] - promo_summary['units_before'])/ promo_summary['units_before']) * 100

promo_summary
```

```python
answer = promo_summary[(promo_summary['IR%'] < 0) &(promo_summary['ISU%'] < 0)]
answer
```

```python
for promo_type, row in answer.iterrows():print(f"{promo_type},{row['IR%']:.2f},{row['ISU%']:.2f}")
```
